### What are Dataframes?
Pandas has a very elegant way of managing data, very much borrowed from the statistical language R, mostly based around dataframes. You can think of dataframes a bit like an Excel table with rows, columns and common operations like sum, average and so on. We can use this on top of our previous work on lists and dictionaries (etc.) more usable and malleable.

To begin with we will create a dataset - in this case a [dictionary](https://github.com/MJMortensonWarwick/IB2AD0_Data_Science_GenerativeAI/blob/main/1_05_dictionaries.ipynb).



```python
orders = {'o10001':{'date':'2024/01/10', 'product':'Hoodie', 'quantity':'1'},
            'o10002':{'date':'2024/01/13', 'product':'Tote bag', 'quantity':'2'},
            'o10003':{'date':'2024/01/14', 'product':'Pencil', 'quantity':'10'},
            'o10004':{'date':'2024/01/15', 'product':'T-shirt', 'quantity':'2'}
}
orders
```

We can convert this to a Dataframe with great ease

```python
import pandas as pd
import numpy as np

orders_df = pd.DataFrame(orders)
orders_df
```

We can even create such outputs from more complex dictionaries, such as a dictionary which includes a nested list:

```python
customers = {'Mark':{'name':'Mark Johnson', 'open_orders':3, 'orders':['o10001', 'o10002', 'o10004']},
             'Katy':{'name':'Katy Hoad', 'open_orders':0, 'orders':[]},
             'Angela':{'name':'Angela Lorenz', 'open_orders':1, 'orders':['o10003']},
             'Bo':{'name':'Bo Kelestyn', 'open_orders':0, 'orders':[]}
}
customers
```

```python
customers_df = pd.DataFrame(customers)
customers_df
```

Pandas dataframes can also be built from various other data structures and sources (including Excel files, CSVs, text files, databases and many more). For example, we can make our dataframe from a set of [lists](https://github.com/MJMortensonWarwick/IB2AD0_Data_Science_GenerativeAI/blob/main/1_04_lists.ipynb):

```python
a = [1, 2, 3, 4]
b = ["a", "b", "c", "d"]
c = [True, False, True, False]

listdf = pd.DataFrame([a, b, c])
listdf
```

### 1. A DataFrame from a dictionary
The following block creates a dictionary where each key is a dog breed, and each value is another dictionary with the breed's country of origin, size and typical lifespan in years.

```python
Tbreeds = {'Labrador Retriever': {'origin': 'Canada', 'size': 'Large', 'lifespan': 12},
          'Border Collie': {'origin': 'United Kingdom', 'size': 'Medium', 'lifespan': 13},
          'Shiba Inu': {'origin': 'Japan', 'size': 'Small', 'lifespan': 14},
          'French Bulldog': {'origin': 'France', 'size': 'Small', 'lifespan': 11}
}

breeds_df = pd.DataFrame(breeds)
breeds_df
```

The outer keys (the breed names) became the **columns**. It's easier to read with one breed per row, so the next block flips the table using `.T` ("transpose").

```python
breeds_df.T
```

### 2. A DataFrame from lists
The following block builds the same kind of table from three lists. Each list becomes one **row**.

```python
breed_names = ['Labrador Retriever', 'Border Collie', 'Shiba Inu', 'French Bulldog']
lifespans = [12, 13, 14, 11]
is_small = [False, False, True, True]

breeds_list_df = pd.DataFrame([breed_names, lifespans, is_small])
breeds_list_df
```

### 3. Testing a missing item
**My prediction:** I ecpect pandas to throw an error because the lists are no longer the same length, or leave a blank where the missing value should be.

**Test A: lists of lists.** Here `is_small` only has three items instead of four.

```python
is_small_short = [False, False, True]

short_list_df = pd.DataFrame([breed_names, lifespans, is_small_short])
short_list_df
```

Pandas did not throw an error. It filled the gap in the last row with `None`, which is how Python represents "no value".

**Test B: a dictionary of lists.** This time each list becomes a **column**. I used `try` and `except` so that, if pandas throws an error, the message is printed instead of stopping the notebook.

```python
try:
    short_dict_df = pd.DataFrame({'breed': breed_names, 'lifespan': lifespans, 'is_small': is_small_short})
    print(short_dict_df)
except ValueError as error:
    print('Error:', error)
```

This time pandas refused and raised an error, because every column must have the same number of rows.

**Test C: a dictionary of dictionaries.** Here the Shiba Inu is missing its "size" entry.

```python
breeds_missing = {'Labrador Retriever': {'origin': 'Canada', 'size': 'Large'},
                  'Shiba Inu': {'origin': 'Japan'}
}

pd.DataFrame(breeds_missing)
```

Pandas matched the values up by their keys and filled the Shiba Inu's missing size with `NaN` ("Not a Number"), pandas' marker for a missing value.

### Conclusion
Whether pandas accepts a missing item depends on how the data is structured:
- **List of lists (each list is a row):** works, and the gap is filled with `None`.
- **Dictionary of lists (each list is a column):** fails with a `ValueError`, as all columns must be the same length.
- **Dictionary of dictionaries:** works, and the gap is filled with `NaN`, because pandas lines values up by their labels.
